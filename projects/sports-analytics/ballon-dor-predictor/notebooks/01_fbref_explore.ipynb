## Soccer Data EDA

```python
import soccerdata as sd
import pandas as pd
```

```python
fb = sd.FBref(leagues="Big 5 European Leagues Combined", seasons="2023-2024")
stats = fb.read_player_season_stats(stat_type="standard")
```

```python
seasons = [f"{y%100:02d}{(y+1)%100:02d}" for y in range(2009, 2025)]
seasons
```

```python
fb = sd.FBref(leagues="Big 5 European Leagues Combined", seasons=seasons)
stats = fb.read_player_season_stats(stat_type="standard")
```

```python
stats.index.get_level_values('season').unique()
```

```python
df = stats.reset_index()
```

```python
df['bdo_season'] = df['season'].str[-2:].astype(int)+2000
```

```python
df[["season", "bdo_season"]].drop_duplicates()
```

```python
df.columns.tolist()
```

```python
cols = {("bdo_season", ""): "season", ("player", ""): "player_name", ("team", ""): "club", ("league", ""): "league", ("pos", ""): "position", ("age", ""): "age", ("Playing Time", "Min"): "minutes", ("Performance", "Gls"): "goals", ("Performance", "Ast"): "assists", ("Per 90 Minutes", "Gls"): "goals_per90"}
```

```python
df_stats = df[list(cols)]
df_stats.shape
```

```python
df_stats = df_stats.set_axis(list(cols.values()), axis=1)
df_stats.columns.tolist()
```

```python
df_stats.head()
```

```python
df_stats["main_position"] = df_stats["position"].str.split(",").str[0]
df_stats['main_position'].value_counts()
```

```python
df_stats = df_stats.dropna(subset=["main_position"])
```

```python
df_stats = df_stats.drop_duplicates()
df_stats[df_stats.duplicated(subset=["season", "player_name", "club"], keep=False)]
```

```python
df_stats.to_csv("../data/raw/fbref_player_stats.csv", index=False)
```
