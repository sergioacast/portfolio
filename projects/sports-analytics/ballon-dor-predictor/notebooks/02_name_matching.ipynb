```python
import pandas as pd

nom = pd.read_csv("../data/raw/ballon_d_or_rankings.csv")
df_stats = pd.read_csv("../data/raw/fbref_player_stats.csv")
```

```python
name_map = {
    "Vinícius Júnior": "Vinicius Júnior",
    "Cesar Azpilicueta": "César Azpilicueta",
    "Eric Abidal": "Éric Abidal",
    "Maicon": "Maicon Sisenando",
    "Fabián Ruiz": "Fabián Ruiz Peña"
}

nom["player_name"] = nom["player_name"].replace(name_map)
```

```python
merged_df = pd.merge(nom, df_stats, on=["player_name", "season"], how="left", indicator=True)
```
