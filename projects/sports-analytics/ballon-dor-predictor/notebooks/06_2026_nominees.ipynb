```python
fb = sd.FBref(leagues='Big 5 European Leagues Combined', seasons="2025-2026")
stats = fb.read_player_season_stats(stat_type='standard')
```

```python
stats.shape
```

```python
stats.index.get_level_values("season").unique()
```

```python
df = stats.reset_index()

df['bdo_season'] = df['season'].str[-2:].astype(int)+2000

cols = {("bdo_season", ""): "season", ("player", ""): "player_name", ("team", ""): "club", ("league", ""): "league", ("pos", ""): "position", ("age", ""): "age", ("Playing Time", "Min"): "minutes", ("Performance", "Gls"): "goals", ("Performance", "Ast"): "assists", ("Per 90 Minutes", "Gls"): "goals_per90", ("nation", ""): "nation"}

df_stats = df[list(cols)].set_axis(list(cols.values()), axis=1)

df_stats["main_position"] = df_stats["position"].str.split(",").str[0]

df_stats = df_stats.dropna(subset=["main_position"])

df_stats = df_stats.drop_duplicates()

df_stats["league"] = df_stats["league"].fillna("GER-Bundesliga")
```

```python
df_stats.to_csv("../data/raw/fbref_player_stats_2026.csv", index=False)
```
